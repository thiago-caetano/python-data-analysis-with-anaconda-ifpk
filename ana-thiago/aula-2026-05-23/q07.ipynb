```python
# Importa biblioteca principal para análise de dados
import pandas as pd

# Importa biblioteca para exibir tabelas no Jupyter
from IPython.display import display

# Importa bibliotecas para limpar nomes das colunas
import re
import unicodedata

# Faz o Pandas mostrar todas as colunas
pd.set_option("display.max_columns", None)

# Faz o Pandas mostrar até 100 linhas
pd.set_option("display.max_rows", 100)

# Link direto do dataset da NASA
url = "https://data.nasa.gov/docs/legacy/meteorite_landings/Meteorite_Landings.csv"

# Lê o dataset direto do link
df = pd.read_csv(url)

# Função para padronizar os nomes das colunas
def limpar_nome_coluna(nome):
    # Converte o nome da coluna para texto
    nome = str(nome)

    # Remove acentos, caso existam
    nome = unicodedata.normalize("NFKD", nome)
    nome = nome.encode("ascii", "ignore").decode("utf-8")

    # Converte para letras minúsculas
    nome = nome.lower()

    # Remove espaços no começo e no final
    nome = nome.strip()

    # Troca espaços e caracteres especiais por "_"
    nome = re.sub(r"[^a-z0-9]+", "_", nome)

    # Remove "_" do começo e do final
    nome = nome.strip("_")

    # Retorna o nome padronizado
    return nome

# Padroniza os nomes das colunas
df.columns = [limpar_nome_coluna(coluna) for coluna in df.columns]

# Converte a coluna year para data
# Valores inválidos serão transformados em NaT
df["year"] = pd.to_datetime(df["year"], errors="coerce")

# Cria uma coluna chamada ano, pegando apenas o ano da data
df["ano"] = df["year"].dt.year

# Função para transformar número em algarismo romano
def numero_para_romano(numero):
    # Lista com os principais valores romanos
    valores_romanos = [
        (1000, "M"),
        (900, "CM"),
        (500, "D"),
        (400, "CD"),
        (100, "C"),
        (90, "XC"),
        (50, "L"),
        (40, "XL"),
        (10, "X"),
        (9, "IX"),
        (5, "V"),
        (4, "IV"),
        (1, "I")
    ]

    # Variável que vai guardar o resultado em romano
    resultado = ""

    # Percorre os valores romanos
    for valor, simbolo in valores_romanos:
        # Enquanto o número for maior ou igual ao valor atual
        while numero >= valor:
            # Adiciona o símbolo romano ao resultado
            resultado += simbolo

            # Subtrai o valor do número
            numero -= valor

    # Retorna o número convertido para romano
    return resultado

# Função para calcular o século com base no ano
def calcular_seculo(ano):
    # Verifica se o ano está vazio
    if pd.isna(ano):
        return "Sem registro"

    # Converte o ano para número inteiro
    ano = int(ano)

    # Calcula o século
    # Exemplo: 1901 a 2000 = século 20
    # Exemplo: 2001 a 2100 = século 21
    seculo_numero = ((ano - 1) // 100) + 1

    # Converte o século para algarismo romano
    seculo_romano = numero_para_romano(seculo_numero)

    # Retorna o século formatado
    return f"Século {seculo_romano}"

# Cria a nova coluna seculo
df["seculo"] = df["ano"].apply(calcular_seculo)

# Mostra algumas colunas para conferir o resultado
display(df[["name", "year", "ano", "seculo"]].head(20))

# Mostra a quantidade de meteoritos por século
contagem_por_seculo = df["seculo"].value_counts().reset_index()

# Renomeia as colunas para ficar mais amigável
contagem_por_seculo.columns = ["seculo", "quantidade"]

# Mostra a contagem por século
display(contagem_por_seculo)
```
