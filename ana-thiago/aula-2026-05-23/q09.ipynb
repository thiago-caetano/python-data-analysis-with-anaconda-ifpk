```python
# Importa biblioteca principal para análise de dados
import pandas as pd

# Importa biblioteca para exibir tabelas no Jupyter
from IPython.display import display

# Importa bibliotecas para limpar nomes das colunas
import re
import unicodedata

# Faz o Pandas mostrar todas as colunas
pd.set_option("display.max_columns", None)

# Faz o Pandas mostrar até 100 linhas
pd.set_option("display.max_rows", 100)

# Link direto do dataset da NASA
url = "https://data.nasa.gov/docs/legacy/meteorite_landings/Meteorite_Landings.csv"

# Lê o dataset direto do link
df = pd.read_csv(url)

# Função para padronizar os nomes das colunas
def limpar_nome_coluna(nome):
    # Converte o nome da coluna para texto
    nome = str(nome)

    # Remove acentos, caso existam
    nome = unicodedata.normalize("NFKD", nome)
    nome = nome.encode("ascii", "ignore").decode("utf-8")

    # Converte para letras minúsculas
    nome = nome.lower()

    # Remove espaços no começo e no final
    nome = nome.strip()

    # Troca espaços e caracteres especiais por "_"
    nome = re.sub(r"[^a-z0-9]+", "_", nome)

    # Remove "_" do começo e do final
    nome = nome.strip("_")

    # Retorna o nome padronizado
    return nome

# Padroniza os nomes das colunas
df.columns = [limpar_nome_coluna(coluna) for coluna in df.columns]

# Converte a coluna de latitude para valor numérico
# Valores inválidos serão transformados em NaN
df["reclat"] = pd.to_numeric(df["reclat"], errors="coerce")

# Função para classificar o hemisfério de acordo com a latitude
def classificar_hemisferio(latitude):
    # Se a latitude estiver vazia, retorna Sem localização
    if pd.isna(latitude):
        return "Sem localização"

    # Latitude maior que zero indica Hemisfério Norte
    if latitude > 0:
        return "Hemisfério Norte"

    # Latitude menor que zero indica Hemisfério Sul
    elif latitude < 0:
        return "Hemisfério Sul"

    # Latitude igual a zero indica Linha do Equador
    else:
        return "Linha do Equador"

# Cria a nova coluna hemisferio
df["hemisferio"] = df["reclat"].apply(classificar_hemisferio)

# Mostra algumas colunas para conferir o resultado
display(df[["name", "reclat", "reclong", "geolocation", "hemisferio"]].head(20))

# Mostra a quantidade de meteoritos em cada hemisfério
contagem_hemisferio = df["hemisferio"].value_counts().reset_index()

# Renomeia as colunas para ficar mais amigável
contagem_hemisferio.columns = ["hemisferio", "quantidade"]

# Mostra a tabela com a contagem
display(contagem_hemisferio)
```
