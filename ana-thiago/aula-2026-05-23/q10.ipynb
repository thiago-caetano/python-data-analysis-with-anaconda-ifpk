```python
# Importa biblioteca principal para análise de dados
import pandas as pd

# Importa biblioteca para exibir tabelas no Jupyter
from IPython.display import display

# Importa bibliotecas para limpar nomes das colunas e textos
import re
import unicodedata

# Faz o Pandas mostrar todas as colunas
pd.set_option("display.max_columns", None)

# Faz o Pandas mostrar até 100 linhas
pd.set_option("display.max_rows", 100)

# Link direto do dataset da NASA
url = "https://data.nasa.gov/docs/legacy/meteorite_landings/Meteorite_Landings.csv"

# Lê o dataset direto do link
df = pd.read_csv(url)

# Função para padronizar os nomes das colunas
def limpar_nome_coluna(nome):
    # Converte o nome da coluna para texto
    nome = str(nome)

    # Remove acentos, caso existam
    nome = unicodedata.normalize("NFKD", nome)
    nome = nome.encode("ascii", "ignore").decode("utf-8")

    # Converte para letras minúsculas
    nome = nome.lower()

    # Remove espaços no começo e no final
    nome = nome.strip()

    # Troca espaços e caracteres especiais por "_"
    nome = re.sub(r"[^a-z0-9]+", "_", nome)

    # Remove "_" do começo e do final
    nome = nome.strip("_")

    # Retorna o nome padronizado
    return nome

# Padroniza os nomes das colunas
df.columns = [limpar_nome_coluna(coluna) for coluna in df.columns]

# Função para extrair o grupo principal da coluna recclass
def extrair_grupo_classe(recclass):
    # Verifica se o valor está vazio
    if pd.isna(recclass):
        return "Sem classe"

    # Converte o valor para texto
    classe = str(recclass).strip()

    # Remove informações depois de vírgula
    # Exemplo: "Iron, IVA" vira "Iron"
    classe = classe.split(",")[0]

    # Remove informações dentro de parênteses
    # Exemplo: "Martian (shergottite)" vira "Martian"
    classe = re.sub(r"\(.*?\)", "", classe)

    # Remove informações depois de hífen
    # Exemplo: "Eucrite-mmict" vira "Eucrite"
    classe = classe.split("-")[0]

    # Remove espaços extras
    classe = classe.strip()

    # Remove números do final da classe
    # Exemplo: "L5" vira "L", "H6" vira "H", "CM2" vira "CM"
    classe = re.sub(r"\d+$", "", classe)

    # Remove espaços extras novamente
    classe = classe.strip()

    # Se depois da limpeza ficar vazio, retorna Sem classe
    if classe == "":
        return "Sem classe"

    # Retorna o grupo principal
    return classe

# Cria a nova coluna grupo_classe
df["grupo_classe"] = df["recclass"].apply(extrair_grupo_classe)

# Mostra algumas linhas para conferir o resultado
display(df[["name", "recclass", "grupo_classe"]].head(30))

# Mostra a quantidade de meteoritos por grupo principal
contagem_grupo_classe = df["grupo_classe"].value_counts().reset_index()

# Renomeia as colunas para ficar mais amigável
contagem_grupo_classe.columns = ["grupo_classe", "quantidade"]

# Mostra a tabela de contagem por grupo principal
display(contagem_grupo_classe.head(30))
```
