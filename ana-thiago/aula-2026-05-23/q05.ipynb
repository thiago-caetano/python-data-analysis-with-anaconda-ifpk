```python
# Importa biblioteca principal para análise de dados
import pandas as pd

# Importa biblioteca para exibir tabelas no Jupyter
from IPython.display import display

# Importa bibliotecas para limpar nomes das colunas
import re
import unicodedata

# Link direto do dataset da NASA
url = "https://data.nasa.gov/docs/legacy/meteorite_landings/Meteorite_Landings.csv"

# Lê o dataset direto do link
df = pd.read_csv(url)

# Função para padronizar os nomes das colunas
def limpar_nome_coluna(nome):
    # Converte o nome da coluna para texto
    nome = str(nome)

    # Remove acentos, caso existam
    nome = unicodedata.normalize("NFKD", nome)
    nome = nome.encode("ascii", "ignore").decode("utf-8")

    # Converte para letras minúsculas
    nome = nome.lower()

    # Remove espaços no começo e no final
    nome = nome.strip()

    # Troca espaços e caracteres especiais por "_"
    nome = re.sub(r"[^a-z0-9]+", "_", nome)

    # Remove "_" do começo e do final
    nome = nome.strip("_")

    # Retorna o nome padronizado
    return nome

# Padroniza os nomes das colunas
df.columns = [limpar_nome_coluna(coluna) for coluna in df.columns]

# Converte a coluna mass_g para valor numérico
# Valores que não puderem ser convertidos serão transformados em NaN
df["mass_g"] = pd.to_numeric(df["mass_g"], errors="coerce")

# Verifica meteoritos com massa igual a zero
meteoritos_massa_zero = df[df["mass_g"] == 0]

# Verifica meteoritos com massa negativa
meteoritos_massa_negativa = df[df["mass_g"] < 0]

# Verifica todos os meteoritos com massa inválida para análise: zero ou negativa
meteoritos_massa_zero_ou_negativa = df[df["mass_g"] <= 0]

# Mostra a quantidade de registros encontrados
print("Quantidade de meteoritos com massa igual a zero:")
print(len(meteoritos_massa_zero))

print("\nQuantidade de meteoritos com massa negativa:")
print(len(meteoritos_massa_negativa))

print("\nQuantidade total de meteoritos com massa zero ou negativa:")
print(len(meteoritos_massa_zero_ou_negativa))

# Se existirem registros com massa zero ou negativa, mostra alguns exemplos
if len(meteoritos_massa_zero_ou_negativa) > 0:
    print("\nExistem meteoritos com massa igual a zero ou negativa.")
    print("Esses registros podem prejudicar a análise porque massa zero ou negativa não faz sentido fisicamente para um meteorito.")
    print("Eles podem distorcer cálculos estatísticos, como média, menor valor, distribuição de massa e gráficos.")

    # Mostra algumas colunas importantes desses registros
    display(
        meteoritos_massa_zero_ou_negativa[
            ["name", "id", "recclass", "mass_g", "fall", "year", "reclat", "reclong", "geolocation"]
        ].head(20)
    )
else:
    print("\nNão existem meteoritos com massa igual a zero ou negativa no dataset.")
    print("Isso indica que, considerando esse critério, a coluna de massa não possui valores fisicamente inválidos.")
```
